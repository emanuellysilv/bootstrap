## 🎯 Entrada de Dados

## 🔀 Condicional: if / else

Permite executar diferentes blocos de código dependendo de uma condição.

```python
idade = int(input("Digite sua idade: "))

if idade >= 18:
    print("Você é maior de idade.")
else:
    print("Você é menor de idade.")
```

## 🔁 Laços de repetição: for

Usamos `for` para repetir ações com base em um iterável (lista, range, etc).

```python
for i in range(1, 6):
    print("Número:", i)
```

### 🧪 Faça você mesmo:

1. Escreva um laço que imprima os múltiplos de 3 entre 1 e 30.


```python
print("Multiplos de 3 entre 1 e 30:")
for numero in range(1, 31):
    if numero % 3 == 0:
        print(numero)
```

## 📚 Listas

Listas armazenam vários elementos em uma única variável.

### Adição:

```python
frutas = []
frutas.append("maçã")
frutas.append("banana")
frutas.append("laranja")
print(frutas)
```

### Remoção:

```python
frutas.remove("maçã")
print(frutas)
```

```python
dias = ['segunda', 'terça', 'quarta', 'quinta', 'sexta', 'sábado', 'domingo']
# Remove o último elemento da lista
dias.pop()
print(dias)

# Remove o elemento de acordo com o índice do elemento
dias.pop(3)
print(dias)
```

### Acesso a elementos:

```python
nomes = ["Ana", "Bruno", "Carlos"]
print(nomes[0])  # Primeiro
print(nomes[-1])  # Último
```

### Uso como iterável:

```python
for fruta in frutas:
    print("Fruta:", fruta)
```

### 🧪 Faça você mesmo:

1. Crie uma lista com 4 nomes.  
2. Adicione um nome.  
3. Remova o segundo nome.  
4. Imprima todos os nomes usando um `for`.


```python
nomes = ["Nivia" , "Bruno", "Nai", "Manu"]
nomes.append("Eddie")
nomes.remove("Manu")

for nome in nomes:
    print(nome)
```

## ⚡ List Comprehensions

Forma rápida de criar listas com base em uma regra.

```python
quadrados = [x**2 for x in range(1, 11)]
print(quadrados)
```

### 🧪 Faça você mesmo:

1. Crie uma lista com os números pares de 0 a 20 usando list comprehension.


```python
numeros = [i for i in range(21) if i % 2 == 0]
print(numeros)
```

Crie um pequeno programa que:

- Peça o nome de 3 alunos e suas notas

```python
nome1 = input("Digite seu nome: ")
nota1 = float(input("Digite sua nota: "))
print()
nome2 = input("Digite seu nome: ")
nota2 = float(input("Digite sua nota: "))
print()
nome3 = input("Digite seu nome: ")
nota3 = float(input("Digite sua nota: "))
print()

alunos = [
    {"nome": nome1, "nota": nota1},
    {"nome": nome2, "nota": nota2},
    {"nome": nome3, "nota": nota3}
]

for aluno in alunos:
    if aluno["nota"] >= 7:

        print()
        print(f"{aluno['nome']} foi aprovado!")
        print("Parabéns!")

    else:

        print()
        print(f"{aluno['nome']} foi reprovado.")
        print("Estude mais!")
```
